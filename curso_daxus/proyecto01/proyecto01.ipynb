```python
from fpdf import FPDF

pdf = FPDF()
pdf.add_page()
pdf.set_font("Arial", size=14)

pdf.cell(50, 10, "Nombre del Proyecto:", border=0)
pdf.cell(0, 10, str(proyecto), ln=True) # ln=True hace un salto de línea al final

pdf.cell(50, 10, "Horas estimadas:", border=0)
pdf.cell(0, 10, str(hora_estimadas), ln=True)

pdf.cell(50, 10, "Valor de la hora:", border=0)
pdf.cell(0, 10, f"${valor_hora}", ln=True)

# Dibujamos una línea para separar el total
pdf.line(10, pdf.get_y(), 100, pdf.get_y()) 

pdf.set_font("Arial", "B", 14) # Negrita para el total
pdf.cell(50, 15, "VALOR TOTAL:", border=0)
pdf.cell(0, 15, f"${valor_total}", ln=True)


pdf.output("Presupuesto.pdf")
print("Formulario creado con éxito.")
```

Hoja PDF formateada

```python
from fpdf import FPDF

class PresupuestoPDF(FPDF):
    def header(self):
        # Configurar color de la marca (Azul suave por ejemplo)
        self.set_fill_color(230, 230, 230)
        self.rect(0, 0, 210, 40, 'F') # Franja superior gris
        
        self.set_font("Arial", "B", 20)
        self.set_text_color(50, 50, 50)
        self.cell(0, 10, "PRESUPUESTO DE PROYECTO", ln=True, align="C")
        self.ln(10)

    def footer(self):
        self.set_y(-15)
        self.set_font("Arial", "I", 8)
        self.set_text_color(128)
        self.cell(0, 10, f"Página {self.page_no()}", align="C")


# --- Crear el PDF con la plantilla ---
pdf = PresupuestoPDF()
pdf.add_page()
pdf.set_margins(20, 20, 20)

# Cuerpo del presupuesto
pdf.ln(20)
pdf.set_font("Arial", "B", 12)
pdf.set_text_color(0)

# Crear una tabla visual simple
def crear_fila(label, valor):
    pdf.set_fill_color(245, 245, 245)
    pdf.set_font("Arial", "B", 11)
    pdf.cell(60, 12, f" {label}", border=1, fill=True)
    pdf.set_font("Arial", "", 11)
    pdf.cell(0, 12, f" {valor}", border=1, ln=True)

crear_fila("Nombre del Proyecto", proyecto)
crear_fila("Horas Estimadas", f"{hora_estimadas} hs")
crear_fila("Valor por Hora", f"${valor_hora}")

# Fila de Total destacada
pdf.ln(5)
pdf.set_fill_color(100, 149, 237) # Azul Cornflower
pdf.set_text_color(255)
pdf.set_font("Arial", "B", 14)
pdf.cell(60, 15, " TOTAL", border=1, fill=True)
pdf.cell(0, 15, f" ${valor_total} USD", border=1, fill=True, ln=True, align='R')

pdf.output("presupuesto_formateado.pdf")
print("PDF con plantilla generado.")
```
