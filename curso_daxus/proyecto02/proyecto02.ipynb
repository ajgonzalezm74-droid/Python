Obteniendo datos para una opcion

```python

import yfinance

ticker= input("Ingrese el codigo de accion: ")
dt_inicial = input("Introduzca la fecha de inicio (aaa-mm-dd): ")
dt_final = input("Ingrese la fecha final (aaaa-mm.dd.): ")

datos = yfinance.Ticker(ticker)
tabla = datos.history(start=dt_inicial, end =dt_final)
tabla
```

Seleccionar solo la columna Cierre (Close)

```python
cierre= tabla.Close
cierre
```

Generar un gráfico lineal (utiliza la biblioteca matplotlib)


```python
#!pip install matplotlib
```

```python
cierre.plot()
```

Generando estadísticas

```python
maxima = round(cierre.max(),2)
minima = round(cierre.min(),2)
valor_medio = round(cierre.mean(),2)

print(maxima)
print(minima)
print(valor_medio)
```

```python
import pyautogui
import pyperclip
import webbrowser
from time import sleep

# --- Datos ---
destinatario = "ajgonzalezm80@gmail.com"
asunto = "Informe de Acciones"


# Mensaje formateado profesionalmente
mensaje = f"""Hola, este es un informe automatizado.

Resumen de análisis:
-------------------------------------------
Acción:          {ticker}
Periodo:         {dt_inicial} al {dt_final}
-------------------------------------------
Cotización Máx:  USD {maxima}
Cotización Mín:  USD {minima}
Valor Medio:     USD {valor_medio}

Atentamente,
Tu Script de Python"""

# --- Ejecución ---

# 1. Abrir Gmail
webbrowser.open("https://mail.google.com/")
sleep(8)  # Espera a que cargue

# 2. Iniciar nuevo correo (Atajo universal de Gmail: 'c' de Compose)
# Nota: Gmail debe tener los "Atajos de teclado" activados en configuración
pyautogui.press('c') 
sleep(3)

# 3. Escribir destinatario (Usando portapapeles para evitar fallos con la @)
pyperclip.copy(destinatario)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")
sleep(1)

# 4. Escribir Asunto
pyperclip.copy(asunto)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")
sleep(1)

# 5. Escribir Cuerpo (Pegar bloque completo es 100x más rápido)
pyperclip.copy(mensaje)
pyautogui.hotkey("ctrl", "v")
sleep(2)

# 6. Enviar (Ctrl + Enter es el estándar)
pyautogui.hotkey("ctrl", "enter")

print("✅ Informe enviado con éxito.")
```

Enviar correo sin abrir gmail

```python
import smtplib
from email.message import EmailMessage

# --- CONFIGURACIÓN ---
mi_correo = "ajgonzalezm74@gmail.com"
mi_password = "jbvz ixvi aneg yxwy" # La contraseña de aplicación
destinatario = "ajgonzalezm80@gmail.com"

# --- EL MENSAJE ---
msg = EmailMessage()
msg['Subject'] = "Asunto del Correo"
msg['From'] = mi_correo
msg['To'] = destinatario
msg.set_content("Hola, este mensaje fue enviado sin abrir el navegador.")

# --- ENVÍO SEGURO ---
try:
    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp: # Conexión segura
        smtp.login(mi_correo, mi_password)
        smtp.send_message(msg)
    print("¡Correo enviado con éxito!")
except Exception as e:
    print(f"Error al enviar: {e}")
```
